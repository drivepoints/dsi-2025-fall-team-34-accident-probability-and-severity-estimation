```python
from src.data_loader import get_nearest_signal_id,load_data
from src.train_volume import train_volume_model
from src.train_risk import train_risk_model
from src.inference import AccidentPredictor
import os
import pandas as pd
```

```python
# Create nearest signal id csv if not created
if not os.path.isfile("data/gdot_with_nearest_signal_within_500m.csv"):
    get_nearest_signal_id()
    print('\n--- gdot_with_nearest_signal_within_500m.csv file created ---')
    
# Load data for model training
if not os.path.isfile("data/df_maneuver.csv"):
    df = load_data()
    print('\n--- df_maneuver.csv created ---')
else:
    df = pd.read_csv("data/df_maneuver.csv")

# Train models if not already created
if not os.path.isfile("models/risk_model.pkl"):
    # Train Volume Model (Model A)
    # This returns the df with 'predicted_volume' added
    df_with_vol = train_volume_model(df)

    # Train Risk Model (Model B)
    train_risk_model(df_with_vol)

    print("\n--- Pipeline Complete. Testing Inference ---")
    
# Inference Test
predictor = AccidentPredictor()
```

```python
# Test Case
vol, risk = predictor.predict(
    hour = 19,
    day_of_week=6, 
    month=8, 
    intersection_type='four-way', 
    maneuver='left'
)

print(f"Predicted Traffic: {vol:.0f}")
print(f"Fatal Risk Probability: {risk:.8f}")
```
